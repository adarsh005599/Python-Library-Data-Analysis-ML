```python
df_country = pd.read_excel('Country-Code.xlsx')
df_country.head(5)
```

```python
df_country.columns
```

```python
df.columns
```

```python
final_df = pd.merge(df, df_country, on='Country Code', how='left')
final_df.head(2)
```

```python
country_name = final_df.Country.value_counts().index
country_name
```

```python
final_val = final_df.Country.value_counts().values
final_val
```

```python
plt.pie(final_val[:3], labels=country_name[:3], autopct='%1.2f%%')
```

from above chart we can observe that major bussiness of zomato is happing on india based about the --94% and remaining to US and UK 
so we can say that indian is the main market for the zomato based startups


```python
final_df.columns
```

Now just to see the more data relationa i am going to use the Aggregate rating, Rating color, Rating text

```python
final_df.groupby(['Aggregate rating', 'Rating color', 'Rating text',]).size().reset_index().rename(columns={0:'Rating count'}).head(5)
```

```python
rating = final_df.groupby(['Aggregate rating', 'Rating color', 'Rating text',]).size().reset_index().rename(columns={0:'Rating count'})
rating 
```

## Observation
1. rating from 4.5 to 4.9 with Dark Green --> Excellent
2. rating from 4.0 to 4.4 with Geen --> Very Good
== and so on from data -------->>>>>>>>
3. at the first about the rating 0 means there are 2148 account have not rate the zomato app at all


```python
plt.figure(figsize=[12,5])
sns.barplot(x='Aggregate rating', y= 'Rating count', data=rating, hue='Rating color', palette=['indigo', 'Red', 'Orange', 'Yellow', 'Green', 'Green'])
plt.title('Rating vs Rating count')
```

# Observation's
1. indigo rated color account holders not rated the zomato at all about the numbers 2148
2. Maximum rating is from 2.7 to 4.1

```python
# vs color with count plot

sns.countplot(x='Rating color', data= rating, palette=['indigo', 'Red', 'Orange', 'Yellow', 'Green', 'Green'])
```

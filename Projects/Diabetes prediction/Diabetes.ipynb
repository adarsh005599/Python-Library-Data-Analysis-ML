```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
```

```python
x = diabetes_data.drop(columns='Outcome', axis=1)
y = diabetes_data['Outcome']
```

```python
print("Diabetes dataset without Outcome: ", x)
```

```python
print('Outcome dataset from the diabetes dataset: ', y)
```

# Data Standardization

```python
scaler = StandardScaler()
```

```python
scaler.fit(x)
```

```python
StandardScaler(copy=True, with_mean=True, with_std=True)
```

```python
Standardized_data = scaler.transform(x) 
print(Standardized_data)
```

```python
x = Standardized_data
y = diabetes_data['Outcome']
```

```python
print(x)
print()
print(y)
```

```python
x_test, x_train, y_test, y_train  = train_test_split(x , y , test_size=0.2,  stratify=y, random_state=2)
```

```python
print(x.shape, x_test.shape, x_train.shape)
```

```python
print(y.shape, y_test.shape, y_train.shape)
```

Traning the data based on labeled data

```python
classification = svm.SVC(kernel='linear')
```

```python
classification.fit(x_train, y_train)
```

Model Evaluation

```python
# accuracy prediction for the training data

x_train_prediction = classification.predict(x_train)
training_data_accuracy = accuracy_score(x_train_prediction, y_train)
```

```python
print(training_data_accuracy)
```

```python
# accuracy prediction for the test data

x_test_prediction = classification.predict(x_test)
test_data_accuracy = accuracy_score(x_test_prediction, y_test)
```

```python
print(test_data_accuracy)
```

# Making the predictive system

```python
# let's take some random input data and see the prectiction
input_data = [(1,85,66,29,0,26.6,0.351,31)]

# now make the input as in array shape okay!
input_data_array = np.asarray(input_data)

# reshape the current array data for one instance
input_data_reshaped = input_data_array.reshape(1,-1)

# Now stend the input data
stend_data = scaler.transform(input_data_reshaped)
print(stend_data)

prediction  = classification.predict(stend_data)

if(prediction[0] == 0):
    print('The person is not diabetic🙂', prediction)
else:
    print('The preson is diabetic🥲', prediction)



```
